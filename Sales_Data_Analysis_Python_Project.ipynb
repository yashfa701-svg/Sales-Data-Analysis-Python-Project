```python
import pandas as pd
data = {"product": ["Laptop", "phone", "Tablets", "Laptop", "phone"],"sales": [80000, 50000, 30000, 90000, 60000]}
df = pd.DataFrame(data)
df
```

```python
total_sales = df["sales"].sum()
total_sales
```

```python
best_product = df.loc[df["sales"].idxmax()]
best_product
```

```python
import matplotlib.pyplot as plt
plt.bar(df["product"], df["sales"])
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Sales Analysis")
plt.show()
```

```python
average_sales = df["sales"].mean()
average_sales
```
